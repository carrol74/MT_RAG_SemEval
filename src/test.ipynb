```python
import json
import os
from pathlib import Path
from datasets import load_dataset, concatenate_datasets
```

```python
DATA_ROOT = "data/raw"
DOMAINS = ["clapnq", "cloud", "fiqa", "govt"]
```

```python
corpus_files = {}
    
# 1. Map domains to their file paths
for domain in DOMAINS:
    file_path = os.path.join(DATA_ROOT, domain, f"{domain}.jsonl")
    if os.path.exists(file_path):
        corpus_files[domain] = file_path
```

```python
corpus_files
```

```python
# 2. Load everything at once
# This creates a DatasetDict: {'finance': dataset, 'tech': dataset...}
raw_datasets = load_dataset("json", data_files=corpus_files)

# 3. Add the 'domain' column and merge
all_datasets = []
for domain, dataset in raw_datasets.items():
    # Add a new column 'domain' with the value (e.g., 'finance')
    ds_with_domain = dataset.add_column("domain", [domain] * len(dataset))
    all_datasets.append(ds_with_domain)

# 4. Concatenate into one master Knowledge Base
final_corpus = concatenate_datasets(all_datasets)
```
